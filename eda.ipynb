# Coffee leaf datasets — EDA tối giản
Kiểm tra phân phối nhãn, kích thước ảnh và ảnh mẫu trước khi train.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

DATA_ROOT = Path('data/classification')
EXT = {'.jpg','.jpeg','.png','.bmp','.webp'}
rows=[]
for p in DATA_ROOT.rglob('*'):
    if p.is_file() and p.suffix.lower() in EXT:
        try:
            with Image.open(p) as im: w,h=im.size
            rows.append({'source':p.relative_to(DATA_ROOT).parts[0], 'label':p.parent.name, 'path':str(p), 'width':w, 'height':h})
        except Exception: pass
df=pd.DataFrame(rows)
print(f'{len(df):,} valid images, {df.label.nunique()} raw labels')
display(pd.crosstab(df.label,df.source, margins=True))
```

```python
fig,ax=plt.subplots(1,2,figsize=(14,4))
df.label.value_counts().plot.bar(ax=ax[0],title='Images per raw label')
df.plot.scatter('width','height',alpha=.25,ax=ax[1],title='Image sizes')
plt.tight_layout()
```

```python
sample=df.groupby('label',group_keys=False).sample(1,random_state=42)
fig,axes=plt.subplots(1,len(sample),figsize=(4*len(sample),4))
axes=[axes] if len(sample)==1 else axes
for ax,(_,r) in zip(axes,sample.iterrows()):
    ax.imshow(Image.open(r.path)); ax.set_title(f'{r.label}\n{r.source}'); ax.axis('off')
plt.tight_layout()
```
